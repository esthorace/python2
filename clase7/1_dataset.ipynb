```python
import matplotlib.pyplot as plt
```

```python
servicio_counts = datos["servicio"].value_counts()
total = sum(servicio_counts)
servicio_cuenta_filtrada = servicio_counts[servicio_counts / total > 0.03]
```

```python
plt.pie(servicio_cuenta_filtrada, labels=servicio_cuenta_filtrada.index, autopct='%1.1f%%')
plt.show()
```
